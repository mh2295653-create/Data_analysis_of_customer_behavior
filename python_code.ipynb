```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x:x .fillna(x.median()))
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
```

```python
#create age columns

labels = ['Young adult','Adult','Middle aged','Senior']
df['age_group'] = pd.qcut(df['age'],q=4,labels = labels)
```

```python
df[['age','age_group']].head(10)
```

```python
#create columns purchase_frequency_days
fequency_mapping = {
    'Fortnightly' : 14,
    'Weekly' : 7,
    'Monthly' : 30,
    'Quarterly' : 90,
    'Bi-weekly' : 14,
    'Annually' : 365,
    'Every 3 month' : 90
}
df['purchase_frequency_days']=df['frequency_of_purchases'].map(fequency_mapping)
```

```python
import pandas as pd
from sqlalchemy import create_engine

# এই সেলটি রান করে কানেকশনটি আবার সচল করে নিন
engine = create_engine(
    "mysql+mysqlconnector://root:Asdfg&12345@localhost:3306/sys"
)
```

```python
# ডাটাবেজ থেকে ডাটা লোড করে দেখা
df = pd.read_sql("SELECT * FROM customer_behavior_table", con=engine)
df.head()
```

```python


# quotes এর ভেতরে আপনার কপি করা পাথটি পেস্ট করুন। সামনে 'r' দিতে ভুলবেন না।
df_customer = pd.read_csv("customer_shopping_behavior.csv")

# ডাটা চেক করার জন্য
df_customer.head()
```

```python
# কাস্টমার ডাটা ফ্রেমটিকে MySQL ডাটাবেজে 'customer_table' নামে সেভ করছি
df_customer.to_sql("customer_shopping_behavior.csv", con=engine, if_exists="replace", index=False)

print("কাস্টমার ডাটা সফলভাবে MySQL-এ আপলোড হয়েছে!")
```
